# 🧠 PARTE 1 – INTEGRAÇÃO CONTÍNUA (CI)

Integração Contínua é uma prática onde **cada alteração no código é automaticamente validada**.

---

## 🔴 Sem CI

- Código integrado manualmente
- Bugs acumulados
- Conflitos grandes
- Deploy arriscado

---

## 🟢 Com CI

- Cada commit é validado automaticamente
- Feedback imediato
- Erros detectados cedo
- Código mais estável

---

## 💡 Ideia central

👉 “Nenhum código deveria entrar na branch principal sem ser validado automaticamente”


# 🔁 PARTE 2 – O QUE ACONTECE APÓS UM PUSH

Fluxo completo:

1. Você faz commit
2. Você faz push
3. GitHub recebe evento
4. GitHub procura workflow
5. Cria um Runner (máquina virtual)
6. Baixa seu código
7. Executa os steps
8. Gera resultado (sucesso ou erro)
9. Runner é destruído

---

## 💡 IMPORTANTE

Nada disso acontece no seu computador.

Tudo acontece no ambiente do GitHub.


# 🧠 PARTE 3 – RUNNER

Runner é:

- Máquina virtual temporária
- Criada automaticamente
- Ambiente limpo
- Não guarda nada após execução

---

## ⚠ CONSEQUÊNCIA

Sempre será necessário:

```bash
pip install -r requirements.txt
```

Porque o ambiente sempre começa vazio.

# 💻 PARTE 5 – PRIMEIRO WORKFLOW (VERSÃO MAIS SIMPLES POSSÍVEL)

Este é o menor pipeline possível.


````python
name: Teste CI Inicial

on: push

jobs:
  teste:
    runs-on: ubuntu-latest

    steps:
      - run: echo "CI funcionando corretamente"
````

# ▶️ EXECUTAR

No terminal:

```bash
git add .
git commit -m "ci: pipeline inicial"
git push
```

---

# 🔎 VERIFICAR

GitHub → Aba **Actions**

Você deve ver:

✔ Workflow rodando  
✔ Job executando  
✔ Step com mensagem  

---

## 💡 O que esse teste prova?

👉 Que o GitHub Actions está funcionando


# 🧠 EXPLICAÇÃO DETALHADA DO WORKFLOW

- **name** → nome da pipeline  
- **on** → evento que dispara  
- **jobs** → conjunto de tarefas  
- **runs-on** → sistema do runner  
- **steps** → comandos executados  
- **run** → comando de terminal  

---

👉 Cada step = comando no terminal


# 💻 PARTE 6 – EVOLUINDO O WORKFLOW (ADICIONANDO CÓDIGO REAL)

Agora vamos evoluir para trabalhar com seu projeto Flask.


````python
name: CI - Flask API

on:
  push:
    branches:
      - main

jobs:
  build:
    runs-on: ubuntu-latest

    steps:
      - name: Clonar repositório
        uses: actions/checkout@v3

      - name: Configurar Python
        uses: actions/setup-python@v4
        with:
          python-version: '3.11'

      - name: Instalar dependências
        run: pip install -r requirements.txt

      - name: Verificar sintaxe
        run: python -m py_compile app.py
````

# 🧠 EXPLICAÇÃO LINHA POR LINHA

### checkout
Baixa o código para o runner

### setup-python
Instala Python

### pip install
Instala dependências

### py_compile
Verifica erro de sintaxe

---

Se qualquer step falhar → pipeline falha
