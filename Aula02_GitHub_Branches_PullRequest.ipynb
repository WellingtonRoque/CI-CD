## Conceito de Branch

Durante o desenvolvimento de um sistema, novas funcionalidades precisam ser implementadas constantemente.

No entanto, alterar diretamente a versão principal do sistema pode gerar instabilidade.

Para resolver esse problema utilizamos branches.

Uma branch é uma linha paralela de desenvolvimento.

**Branch principal:**

`main`

**Exemplo de branch de funcionalidade:**

main  
└ feature-cadastro-livros



# Criando uma branch

Criar nova branch:

`git checkout -b feature-cadastro-livros`



## Commit 6 — Criar rota de cadastro de livros


Adicionar nova rota:

````python
@app.route("/cadastro-livro")
def cadastro_livro():
    return "Página de cadastro de livros"

````

`git add .`  
`git commit -m "Cria rota de cadastro de livros"`



## Commit 7 — Melhorar descrição da rota

Alterar função:


````python
@app.route("/cadastro-livro")
def cadastro_livro():
    return "Formulário de cadastro de livros"

````

`git add .`  
`git commit -m "Melhora descrição da rota de cadastro"`



## Enviando branch para GitHub

Enviar branch para o repositório remoto:

`git push origin feature-cadastro-livros`



## Criando Pull Request no GitHub

Pull Request é um pedido para integrar alterações de uma branch em outra.

feature → main

Após enviar a branch para o GitHub, abrir o repositório no navegador.

**Compare & Pull Request**

Clique em:

**Compare & Pull Request**



### Configuração do Pull Request

O GitHub mostrará a seguinte configuração:

**base: main**
**compare: feature-cadastro-livros**

Isso significa que as alterações da branch:

**feature-cadastro-livros**

serão integradas à branch:

**main**



### O que acontece ao clicar em Compare & Pull Request

O GitHub mostra:

- commits realizados
- arquivos modificados
- diferenças no código

Esse processo é chamado de Code Review.



### Criando o Pull Request

**Título:**

Adiciona funcionalidade de cadastro de livros

**Descrição:**

Implementa nova rota /cadastro-livro no sistema Flask

**Clique em:**

Create Pull Request



## Realizando Merge

Após revisar as alterações,

**Clique em:**

Merge Pull Request

**Depois:**

Confirm Merge



## Atualizar projeto local

Depois do merge, voltar ao terminal:

`git checkout main`  
`git pull`



## Conflito de Merge

Conflito ocorre quando duas branches alteram o mesmo trecho do código.



### Gerando conflito

Criar primeira branch:

`git checkout main`  
`git checkout -b feature-home-v1`

Alterar:



````python
@app.route("/")
def home():
    return "Sistema de Biblioteca Online"

````

**Registrar commit:**

`git add .`  
`git commit -m "Atualiza título da home"`  
`git push origin feature-home-v1`



**Criar segunda branch:**

git checkout main  
git checkout -b feature-home-v2

Alterar:

````python
@app.route("/")
def home():
    return "Sistema de Gestão de Biblioteca"
````

**Registrar commit:**


git add .  
git commit -m "Atualiza título da página inicial"  
git push origin feature-home-v2



# Gerando o conflito no GitHub


Realizar merge da branch:

`feature-home-v1 → main`

Depois tentar criar Pull Request:

`feature-home-v2 → main`

O GitHub exibirá a mensagem:

This branch has conflicts that must be resolved

## Resolvendo Conflito


Ao clicar em Resolve conflicts, o GitHub exibirá o código assim:

<<<<<<< HEAD
Sistema de Biblioteca Online

`=======`
Sistema de Gestão de Biblioteca

`>>>>>>>` feature-home-v2

Significado:

HEAD → versão atual da branch main

`=======`
separador das versões

`>>>>>>>` feature-home-v2 → versão da outra branch

**Escolher a versão final:**


Sistema de Gestão de Biblioteca Online


**Depois clicar em:**

Mark as resolved

Commit merge

Atualizar projeto local

Depois do merge, voltar ao terminal:


```
git checkout main
git pull
```
