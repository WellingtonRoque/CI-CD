# 🧠 Parte 1 – Git x GitHub

## 🔍 Git

Git é um sistema de controle de versão distribuído.

Ele permite:

- Registrar alterações  
- Criar histórico  
- Criar branches  
- Controlar versões localmente  

Git funciona na sua máquina.

---

## 🌐 GitHub

GitHub é uma plataforma online que hospeda repositórios Git.

Ele adiciona:

- Interface visual  
- Colaboração em equipe  
- Pull Requests  
- Controle de permissões  
- Integração com CI (GitHub Actions)  

👉 Git controla versões.  
👉 GitHub gerencia colaboração e automação.

# 🌿 Parte 3 – O que é Branch?

Branch significa ramificação.

Imagine o projeto como uma árvore:

main  
 ├── feature/login  
 ├── feature/usuarios  
 └── fix/bug-auth  

Cada branch permite desenvolver isoladamente e evitar quebrar a versão principal.


## ❗ Por que não trabalhar direto na main?

Porque:

- Pode quebrar o sistema  
- Pode afetar produção  
- Pode disparar CI indevidamente  
- Pode gerar conflitos graves  

Em empresas, a branch main é protegida.


```python
git checkout -b feature/usuarios
```

✅ Verificação:

Execute:

git branch

# 🔁 Parte 5 – Pull Request (Conceito Completo)

Pull Request NÃO é apenas um botão de merge.

Ele é um processo formal de integração de código.

Ele envolve:

- Comparação entre branches  
- Revisão técnica  
- Discussão  
- Aprovação  
- Execução de testes automatizados  

👉 Em ambientes com CI/CD, o PR dispara automaticamente a pipeline.


## 📊 O que acontece quando abrimos um PR?

1. GitHub compara main e feature  
2. Mostra diferenças linha por linha  
3. Permite comentários  
4. Pode exigir aprovação  
5. Pode bloquear merge se CI falhar  

Pull Request é controle de qualidade.


💻 PRÁTICA:

1. Acesse o repositório no GitHub  
2. Clique em "Compare & pull request"  
3. Escreva título claro  
4. Descreva o que foi feito  
5. Clique em "Create pull request"  

Antes de fazer merge:

Observe a aba "Files changed".  
Analise as diferenças.


Após o merge:


```python
git checkout main
git pull origin main
```

✅ Verificação Final:

Execute:

git log --oneline

O commit da feature agora aparece na main?
